# APIs: requêtes HTTP

## Imports

```python
import json
import os
import requests
from dotenv import load_dotenv

load_dotenv()
```

## Recherche de journaux historiques avec Europeana. Lien : https://www.europeana.eu

```python
API_KEY = os.getenv("EUROPEANA_API_KEY")
search_term = "Belgian newspapers"

if not API_KEY:
    raise ValueError("La clé EUROPEANA_API_KEY doit être définie dans le fichier .env")

headers = {"X-Api-Key": API_KEY}
url = "https://api.europeana.eu/record/v2/search.json"

data = {
    "query": search_term,
    "theme": "newspaper",
    "rows": 20,
    "profile": "rich"
}

resp = requests.get(url, params=data, headers=headers)

try:
    json_list = json.loads(resp.text)
except json.JSONDecodeError:
    print(resp.text)
    json_list = {"totalResults": 0, "items": []}
```

## Afficher les résultats

https://restcountries.com/

```python
print(f"Nombre total de résultats : {json_list['totalResults']}")

for item in json_list["items"]:
    title = item.get("title", ["Titre inconnu"])[0]
    country = item.get("country", ["Pays inconnu"])[0]
    provider = item.get("dataProvider", ["Institution inconnue"])[0]
    link = item.get("guid", "")

    print(title)
    print(f"Pays : {country}")
    print(f"Institution : {provider}")
    print(f"Lien : {link}")
    print()
```
